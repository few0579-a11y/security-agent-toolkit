## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5 · 함수 — 여러 줄의 일에 이름을 붙인다</mark>

함수는 여러 줄의 코드를 묶어 이름을 붙인 것입니다. 이름을 부르면 묶인 코드가 실행됩니다.

### 왜 함수가 필요한가

1. 오전에 만든 「실패 계정 찾기」는 열 줄쯤 됩니다. 다른 로그에도 쓰려면 그 열 줄을 통째로 복사해야 합니다.
2. 복사해 두면 고칠 일이 생겼을 때 복사한 곳을 전부 찾아다니며 고쳐야 합니다. 하나라도 빠뜨리면 결과가 달라집니다.
3. 코드를 한 번만 묶어 이름을 붙여 두고, 필요할 때마다 그 이름을 부르면 됩니다. 고칠 곳도 한 군데뿐입니다.

라면 끓이는 순서를 종이에 적어 「라면 끓이기」라고 제목을 달아 두는 것과 같습니다. 다음부터는 순서를 다시 적지 않고 제목만 말하면 됩니다.

### 이 시간에 나오는 말

| 말 | 뜻 | 예 |
|---|---|---|
| 함수 | 여러 줄의 코드를 묶어 이름을 붙인 것 | `say_hello` |
| 정의한다 | 그 묶음을 적어 두는 일. `def`로 시작한다 | `def say_hello():` |
| 호출한다 | 이름을 불러 실제로 실행하는 일 | `say_hello()` |
| 매개변수 | 함수가 받아 쓸 값에 붙인 이름 (괄호 안) | `def show_alert(user):`의 `user` |
| 인자 | 부를 때 실제로 넘기는 값 | `show_alert("kim01")`의 `"kim01"` |
| 반환값 | 함수가 돌려주는 결과 | `return count`의 `count` |

### 함수를 만드는 규칙 다섯 가지

1. `def`로 시작하고 한 칸 띄운 뒤 이름을 씁니다.
2. 이름 뒤에는 괄호 `()`, 줄 끝에는 콜론 `:`을 붙입니다. 하나라도 빠뜨리면 `SyntaxError`가 납니다.
3. 함수가 하는 일은 그 아래에 네 칸 들여씁니다. 들여쓰기가 끝나는 곳이 함수의 끝입니다.
4. 이름은 무엇을 하는지 알 수 있게 짓습니다. 소문자와 밑줄을 씁니다 — `count_failed`, `show_alert`. (이름 규칙은 8교시에서 다시 다룹니다.)
5. 정의가 먼저, 호출이 나중입니다. 부를 때도 괄호를 꼭 붙입니다.

<div style="background:#e8f5e9; color:#17351f; padding:12px 16px; border-radius:8px"><strong>🔎 코드를 읽는 법</strong><br>`def` 줄은 ‘이런 일을 이 이름으로 부르겠다’고 적어 두는 것입니다. 적기만 하면 아무 일도 일어나지 않습니다. 이름 뒤에 괄호를 붙여 불러야 실행됩니다.</div>

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.1 여러 줄의 일에 이름을 붙이고 싶을 때</mark>

`def 이름():` 아래 들여쓴 줄이 함수가 하는 일입니다. 부를 때마다 한 번씩 실행됩니다.

````python
def say_hello():  # say_hello라는 이름으로 적어 둔다 (아직 실행 안 됨)
    print("안녕하세요")

say_hello()  # 부르는 순간 실행된다
say_hello()  # 또 부르면 또 실행된다
````

위 코드는 이런 순서로 돕니다.

1. `def say_hello():` — 파이썬이 「이 묶음을 say_hello라고 부르기로 한다」고 기억만 합니다. 아직 아무것도 출력되지 않습니다.
2. `say_hello()` — 이름을 부르는 순간 묶음 안으로 들어가 `print`를 실행합니다.
3. 묶음이 끝나면 부른 자리로 돌아옵니다. 그래서 다음 줄이 이어서 실행됩니다.

부른 횟수만큼 실행됩니다. 두 번 불렀으니 두 줄이 나옵니다.

> ⚠ 이렇게 쓰면 안 됩니다 — 이 코드는 실행하지 않습니다.
>
> ```python
> say_hello   # 괄호가 없다
> ```
>
> 에러는 나지 않지만 함수가 실행되지 않습니다. 코랩에서는 `<function say_hello at 0x…>` 같은 글자만 나옵니다. 부를 때는 반드시 괄호까지 씁니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.3 결과를 돌려받고 싶을 때</mark>

`return`은 함수가 만든 값을 부른 곳으로 돌려줍니다. 돌려받은 값은 변수에 담아 다음 일에 씁니다.

````python
def add(a, b):
    return a + b  # 결과를 부른 곳으로 돌려준다

result = add(3, 5)  # 돌려받은 8이 result에 담긴다
print(result)
````

> 🤔 이런 의문이 들 수 있습니다 — `print`와 `return`은 뭐가 다른가요?
>
> 1. `print`는 화면에 보여 주기만 합니다. 그 값을 변수에 담을 수 없습니다.
> 2. `return`은 값을 돌려줍니다. 받은 쪽에서 담아 두고 계산을 이어 갈 수 있습니다.
> 3. 그래서 결과를 다음 일에 쓸 함수는 `return`으로 끝냅니다.

1. `return`을 만나면 함수는 그 자리에서 끝납니다. 아래에 코드가 더 있어도 실행되지 않습니다.
2. `return`이 없는 함수는 `None`(아무것도 없음)을 돌려줍니다. 결과가 `None`으로 나오면 `return`을 빠뜨린 것입니다.
3. 화면에 보여 주기만 하면 되는 함수는 `print`로 끝내도 됩니다. 결과를 다음 일에 써야 하면 `return`으로 끝냅니다.

````python
logs = [
    {"result": "FAIL"}, {"result": "OK"},
    {"result": "FAIL"}, {"result": "FAIL"},
]

def count_failed(logs):  # 오전 2-4의 코드를 함수로 묶었다
    fail_count = 0
    for log in logs:
        if log["result"] == "FAIL":
            fail_count = fail_count + 1
    return fail_count  # 반복이 끝난 뒤 돌려준다

total = count_failed(logs)
print(f"실패 {total}건")
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.5 오전 코드를 통째로 함수에 넣고 싶을 때</mark>

오전 3교시의 Counter와 기준값 코드를 함수로 묶으면, 어떤 목록이든 한 줄로 확인할 계정을 찾습니다.

````python
from collections import Counter  # 항목별 횟수를 세는 도구

def find_suspects(failed_users, threshold=2):
    counts = Counter(failed_users)
    suspects = []
    for user, count in counts.items():
        if count >= threshold:
            suspects.append(user)
    return suspects

print(find_suspects(["kim01", "kim01", "lee02", "kim01"]))
print(find_suspects(["kim01", "kim01", "lee02", "kim01"], 5))  # 기준을 바꿔 다시 확인
````

````python
with open("sample_logs.csv", encoding="utf-8") as f:
    for line in f:
        print(line.strip())  # 줄 끝의 \n을 지우고 출력
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.1 자주 쓰는 함수를 따로 모아 두고 싶을 때</mark>

함수만 담은 파일을 모듈이라고 합니다. 저장하는 방법은 4.8과 6.6에서 쓴 `%%writefile` 그대로입니다.

````python
%%writefile utils.py
from collections import Counter

def find_suspects(failed_users, threshold=2):
    counts = Counter(failed_users)
    suspects = []
    for user, count in counts.items():
        if count >= threshold:
            suspects.append(user)
    return suspects
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.4 글자로 적힌 시각에서 시만 꺼내고 싶을 때</mark>

`"03:11"`은 글자라서 시와 분을 따로 꺼낼 수 없습니다. `strptime`으로 시각 값으로 바꾸면 `.hour`로 시만 꺼냅니다.

````python
from datetime import datetime

time_text = "03:11"
log_time = datetime.strptime(time_text, "%H:%M")  # 글자를 시각 값으로

print(log_time.hour)    # 3
print(log_time.minute)  # 11
````

1. `strptime`의 두 번째 값은 「이 글자가 어떤 모양인지」 알려 주는 것입니다. `"03:11"`은 시와 분이라서 `"%H:%M"`입니다.
2. 모양이 다르면 오류가 납니다. `"09:01:30"`처럼 초까지 있으면 `"%H:%M:%S"`로 적습니다.

<table width="100%" bgcolor="#A5D6A7"><tr><td bgcolor="#A5D6A7"><h3><font color="#1B5E20">✍️ 문제 8-3 · 이름을 규칙에 맞게 고치기</font></h3></td></tr></table>

아래 코드는 잘 돌아가지만 이름이 관례에 맞지 않습니다. 이름 세 개를 고쳐 같은 결과가 나오게 하시오.

| 지금 이름 | 무엇인가 | 어떻게 고치나 |
|---|---|---|
| `t` | 한 번 정하고 안 바꾸는 기준값 | 상수 — 전부 대문자 |
| `fu` | 실패 계정 목록 | 변수 — 알아볼 수 있는 이름 |
| `CountFail` | 횟수를 세는 함수 | 함수 — 소문자와 밑줄 |

| | |
|---|---|
| 주어지는 값 | `t` · `fu` · `CountFail` 을 쓰는 코드 |
| 🎯 나와야 하는 결과 | `확인 필요: kim01` |

<details>
<summary>▸ 힌트 보기</summary>

1. `t`는 한 번 정하고 안 바꾸는 기준값입니다. 상수라서 전부 대문자로 `THRESHOLD`가 좋습니다.
2. `fu`는 실패 계정 목록입니다. `failed_users`처럼 알아볼 수 있게 씁니다.
3. `CountFail`은 함수입니다. 소문자와 밑줄로 `count_fail`처럼 씁니다.
4. 이름을 바꿀 때는 그 이름을 쓰는 곳도 같이 바꿉니다. 한 군데만 바꾸면 `NameError`가 납니다.

그래도 막히면 맨 아래 정답 8-3 셀의 「코드 표시」를 누릅니다.

</details>

````python
THRESHOLD = 2
failed_users = ["kim01", "kim01", "lee02"]

def count_fail(users, name):
    count = 0
    for user in users:
        if user == name:
            count = count + 1
    return count

if count_fail(failed_users, "kim01") >= THRESHOLD:
    print("확인 필요: kim01")
````
